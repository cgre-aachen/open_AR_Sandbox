```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
im = np.random.random((20,20))

x = np.linspace(-0.5, 1, 40)
y = np.linspace(0,2,30)

XX, YY = np.meshgrid(x,y)
ZZ = np.cos(XX) + np.sin(YY)
```

```python
fig, ax = plt.subplots()
ax.contour(ZZ)
img = ax.imshow(im)
```

```python
img.set_data(np.ones_like(im))
```

```python
fig.canvas.draw_idle()
```
